```python
import os
from PIL import Image
from PIL.ExifTags import TAGS

image_file = "i4.jpg"

try:
    image = Image.open(image_file)
    print("ya")
except IOError:
    pass

exif = {}

for tag, value in image._getexif().items():
    if tag in TAGS:
        exif[TAGS[tag]] = value

try:
    if "Copyright" in exif:
        print("Image is Copyrighted, by ", exif["Copyright"])
except KeyError:
    pass

print()
print("Displaying metadata of the image: \n")
print(exif)
```

```python
import hashlib
from PIL.PngImagePlugin import PngInfo

name = "Arnav Menon R-106119017"

infolist = name + exif.get("Make") + exif.get("Model") + exif.get("DateTimeOriginal")
hashstring   = hashlib.sha256(infolist.encode()).hexdigest()

cpinfo = PngInfo()
cpinfo.add_text("Copyright", hashstring)
image.save("i4.png", pnginfo=cpinfo)

image = Image.open("i4.png")
print(image.text)
```
