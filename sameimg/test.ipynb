```python
import hashlib as hl
```

```python
def file_hash(file_path):
    sha256 = hl.sha256()
    with open(file_path, "rb") as f:
        while True:
            data = f.read()
            if not data:
                break
            sha256.update(data)

    return sha256.hexdigest()
```

```python
def test1():
    a = file_hash("i1.jpeg")
    b = file_hash("i2.jpeg")

    if a == b:
        print("Same")
    else:
        print("Different")

test1()
```
